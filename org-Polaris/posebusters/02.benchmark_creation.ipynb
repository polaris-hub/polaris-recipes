# Dataset and Benchmark creation with `Polaris`
The first step of creating a benchmark is to set up a standard dataset which allows accessing the curated dataset (which has been demonstrated in <07_01_polaris_adme-fang-1_data_curation.ipynb>), and all necessary information about the dataset such as data source, description of endpoints, units etc. 

```python
%load_ext autoreload
%autoreload 2

import os
import sys
import pathlib
import numpy as np

import datamol as dm

# polaris benchmark
from polaris.benchmark import (
    SingleTaskBenchmarkSpecification,
    MultiTaskBenchmarkSpecification,
)

# polaris hub
from polaris.utils.types import HubOwner

# utils
root = pathlib.Path("__file__").absolute().parents[2]
os.chdir(root)
sys.path.insert(0, str(root))
```

```python
org = "polaris"
gcp_root = f"gs://polaris-public/polaris-recipes/org-{org}/posebusters"

owner = HubOwner(slug="polaris", type="organization")

BENCHMARK_DIR = f"{gcp_root}/benchmarks"
```

## Load existing dataset object

```python
# Get the owner and organization
import polaris as po

dataset = po.load_dataset(path="polaris/posebusters-v1")
```

```python
dataset
```

## Benchmark creation with `Polaris`
PoseBuster dataset is a testset which is pre-defined in [paper](https://pubs.rsc.org/en/content/articlelanding/2024/sc/d3sc04185a). The training set is PDBbinder which cannot be added here due to license restriction.


```python
dataset.table.head(5)
```

## Create a Benchmark as in published paper


```python
train_split = []
test_split = list(range(dataset.table.shape[0]))
```

## Single task on ligand pose prediction

```python
benchmark_name = "posebusters_v1"

benchmark = SingleTaskBenchmarkSpecification(
    name=benchmark_name,
    dataset=dataset,
    target_cols=["ligand"],
    target_types={"ligand": "docking"},
    input_cols=["protein"],
    split=(train_split, test_split),
    main_metric="rmsd_coverage",
    metrics=["rmsd_coverage"],
    tags=["Docking", "SingleTask"],
    owner="polaris",
    description=f"Docking task benchmark for 428 proteins and ligands.",
)
```

```python
benchmark
```

```python
# Upload to polaris hub
benchmark.upload_to_hub(owner=owner, access="private")
```

### Code example for evaluation with 

```python
# load the benchmark
benchmark = po.load_benchmark("polaris/posebusters_v1")
```

```python
# make predictions
# each data point is the docked ligand conformer
# here we use same ligand in dataset for demonstration purpose
predictions = np.array(
    [
        benchmark.dataset.get_data(row=i, col="ligand")
        for i in range(benchmark.dataset.n_rows)
    ]
)
```

```python
# perform evaluation
res = benchmark.evaluate(predictions)
res
```
